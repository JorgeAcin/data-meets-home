```python
import pandas as pd
import time

search_url = "https://api.idealista.com/3.5/es/search"
params = {
    "operation": "sale",  # Venta de inmuebles
    "propertyType": "homes",  # Solo viviendas
    "center": "39.007498, -0.165308",  # Coordenadas del centro de Valencia
    "distance": 3500,  # Radio en metros
    "maxItems": 50,  # Máximo permitido por consulta
    "numPage": 1,  # Empezamos en la página 1
}

# Encabezados de la solicitud
headers = {
    'Authorization': f'Bearer {token}',
    'Content-Type': 'application/x-www-form-urlencoded'
}

# Lista para guardar todas las propiedades
all_properties = []
# Bucle para obtener todas las páginas
while True:
    try:
        response = requests.post(search_url, headers=headers, data=params)
        data = response.json()

        if "elementList" not in data or len(data["elementList"]) == 0:
            break  # Si no hay más resultados, salimos del bucle

        all_properties.extend(data["elementList"])  # Guardamos los resultados

        print(f"Página {params['numPage']} descargada, {len(data['elementList'])} propiedades obtenidas.")

        params["numPage"] += 1  # Pasamos a la siguiente página

        time.sleep(1)  # Esperamos 1 segundo entre peticiones para evitar bloqueos
    except Exception as e:
        print(f"Error: {e}")
        break

print(f"Se han obtenido {len(all_properties)} propiedades en total.")

# Crear un DataFrame con todas las propiedades
df = pd.DataFrame(all_properties)

# Guardar el DataFrame filtrado en un nuevo archivo CSV
df.to_csv("propiedades_valencia.csv", index=False)

# Mostrar el número de filas filtradas
print(f"Se han obtenido {len(df)} propiedades en el municipio de Valencia.")
```
